# **Student Information dictionary**

```python
# Initialize the main dictionary to act as our database
# We use Student IDs as unique keys, and nested dictionaries for their data
student_db = {}
```

```python
def add_student(student_id, name, age, major):
    """Adds a new student to the dictionary."""
    if student_id in student_db:
        print(f"Error: Student ID '{student_id}' already exists.")
    else:
        student_db[student_id] = {
            "Name": name,
            "Age": age,
            "Major": major
        }
        print(f"Success: Student '{name}' added.")
```

```python
print("--- Adding Records ---")
add_student("ID001", "Alice Smith", 20, "Data Science")
add_student("ID002", "Bob Jones", 22, "Computer Engineering")
```

```python
def search_student(student_id):
    """Searches for a student by their ID."""
    if student_id in student_db:
        print(f"Record Found for '{student_id}': {student_db[student_id]}")
        return student_db[student_id]
    else:
        print(f"Not Found: No student matches ID '{student_id}'.")
        return None
```

```python
print("\n--- Searching Records ---")
search_student("ID001")
search_student("ID999") # Testing an invalid search
```

```python
def update_student(student_id, field, new_value):
    """Updates a specific field of an existing student's record."""
    if student_id in student_db:
        if field in student_db[student_id]:
            student_db[student_id][field] = new_value
            print(f"Success: Student '{student_id}' updated {field} to '{new_value}'.")
        else:
            print(f"Error: Field '{field}' does not exist in the record.")
    else:
        print(f"Error: Cannot update. Student ID '{student_id}' not found.")
```

```python
print("\n--- Updating Records ---")
update_student("ID001", "Age", 21)
update_student("ID002", "Major", "Artificial Intelligence")
```

```python
print("\n--- Final Database State ---")
for s_id, info in student_db.items():
    print(f"{s_id}: {info}")
```
